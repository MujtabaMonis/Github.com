```python
payment_counts = df["payment_method"].value_counts()
print(payment_counts)
```

```python
plt.figure(figsize=(6,6))
plt.pie(payment_counts, labels=payment_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Payment Method Distribution")
plt.show()
```

```python
# List of months in order
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']

df['Month'] = pd.Categorical(df['Month'], categories=month_order, ordered=True)

monthwise_sales = df.groupby('Month')['total'].sum().sort_index()

# Plotting
plt.figure(figsize=(20,5))
plt.plot(monthwise_sales.index, monthwise_sales.values, marker='o', linestyle='-', color='b')
plt.title('Monthwise Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

```python
# Example mapping, replace with your city names
city_map = {1: 'Karachi', 2: 'Lahore', 3: 'Islamabad', 4: 'Multan', 5: 'Sukkur', 6: 'Quetta'}
df['City'] = df['City'].map(city_map)
```

```python
import seaborn as sns

city_sales = df.groupby('City')['total'].sum().sort_values()

plt.figure(figsize=(8,5))
sns.barplot(x=city_sales.values, y=city_sales.index, palette='viridis')
plt.title('Citywise Sales')
plt.xlabel('Total Sales')
plt.ylabel('City')
plt.show()
```

```python
monthly_category = df.groupby(['Month','category'])['total'].sum().unstack()

monthly_category.plot(kind='bar', figsize=(12,6))
plt.title('Monthly Sales by Category')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.legend(title='Category')
plt.show()
```

Electronics and Accessories show peak sales in the first half of the year, while Fashion remains relatively steady with a few mid-year spikes. Overall sales for all categories decline after June with mild recovery toward yearend.
